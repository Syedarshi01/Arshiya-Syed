Operators in Python:

1. Arithmetic Operators
2. Assignment Operators
3. Comparison Operators
4. Logical Operators
5. Identity Operators
6. Membership Operators


```python
#1. Arithmetic Operators
n1 = 10
n2 = 2

#Addition
n3 = n1 + n2
print('Add:',n3)

#Subtraction
n4  = n1 - n2
print('Sub:',n4)

#Multiplication
n5 = n1 * n2
print('Multiply:',n5)

#Division
n6 = n1 / n2
print('Divide:',n6)

#Floor Division
n7 = n1 // n2
print('Floor Div:',n7)

#Modulus
n8 = n1 % n2
print('Mod:',n8)

#Exponent
n9 = n1 ** n2
print('Exponent:',n9)


```

2. Assignment Operators:

=

+=

-=

*=

/=

%=

//=

```python
a = 4
print('b:',a)

a += 6
print('c:',a)

a -= 2
print('d:',a)

a *= 6
print('e:',a)

a **= 6
print('f:',a)

a /= 6
print('g:',a)

a %= 6
print('h:',a)

a //= 6
print('i:',a)
```

3. Comparison Operators

```python
a = 6
b = 8

print(a == b)   #equal to
print(a != b)   #not equal to
print(a > b)    #greater than
print(a < b)    #less than
print(a >= b)   #greater than or equal to
print(a <= b)   #less than or equal to
```

4. Logical Operators:

and

or

not

```python
a = 45

b = (a > 40) and (a <50)
print(b)

c = (a > 40) or (a <50)
print(c)

d = not((a > 40) and (a < 50))
print(d)

e = not((a < 40) or (a < 50))
print(e)
```

5. Identity Operator:

Is

Is not

```python
a = 99
b = 99

print(a is b)
print(a is not b)

c = 5
d = 8

print(c is d)
print(c is not d)
```

6. Membership Operators:

In

Not in


```python
a = 4
b = 90
l = [1,2,7,8,9,90]

print(a in l)
print(a not in l)
print(b in l)
print(b not in l)
```
