# Basics Datatypes in Python:

  1. Integer
  2. Floating
  3. Boolean
  4. Complex
  5. String

```python
# As we already know about integers and floating, we will directly go for other types.
#3. Boolean:
a = True
b = False
print(a)
print(b)
print(type(a))
print(type(b))
```

```python
a = 4 > 7
b = 4 < 7
print(a)
print(b)
print(type(a))
print(type(b))
```

```python
# Conversion of Datatypes:
# int to float:
a = 4
print(type(a))
print(a)
a = float(a)
print(type(a))
print(a)
```

```python
# float to int:
a = 4.5
print(type(a))
print(a)
a = int(a)
print(type(a))
print(a)
```

```python
#4. Complex:
c = 1 + 1j
print(c)
print(type(c))
```

```python
#As we also know about strings let's do some more operations on it.
str1 = ('My Collaboration')
print(str1)
print(type(str1))
str2 =  'Arshiya Syed'
print(str2)
print(type(str2))
```
