Factorial of a Number:

it is the product of all positive integers less than or equal to the given integer.
Ex:
Factorial of 6: 6 * 5 * 4 * 3 * 2 * 1 = 720.

```python
num = int(input('Enter the Number:'))

factorial = 1

if (num == 0):
  print('The factorial of 0 is: 1')
else:
  for i in range(1, num + 1):    #range(1,7) ---> 1,2,3,4,5,6
    factorial = factorial * i
print('The factorial of',num,'is: ', factorial)
```

Factorial function

```python
def fac_val(num):

  factorial = 1

  if num == 0:
    return factorial

  else:
    for i in range(1, num + 1):
      factorial = factorial * i
    return factorial
```

```python
print(fac_val(6))
```

```python
print(fac_val(34))
```
