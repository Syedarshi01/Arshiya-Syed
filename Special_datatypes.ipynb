List

```python
#List should be given in square brackets only
l1 = [1,2,3,4]
print(l1)
print(type(l1))
```

```python
#list can have multiple datatypes
l2 = [1,1,2,4.56,True,3.9j,'arshiya']
print(l2)
print(type(l2))
```

```python
#lists are mutable, means they can change
l2 = [1,1,2,4.56,True,3.9j,'arshiya']
l2.append(98)
print(l2)
```

```python
#Printing of elements using their indexes from list
l2 = [1,1,2,4.56,True,3.9j,'arshiya']
print(l2[0])
print(l2[6])
```

```python
#List can have duplicates
l2 = [1,1,2,4.56,True,3.9j,'arshiya']
print(l2)
l2.append(1)
l2.append(99)
l2.append(99)
print(l2)
print(len(l2))
```

```python
#Initiating an empty list
l3 = []
print(l3)
```

```python
l3.append('sakhi')
print(l3)
```

```python
#Deleting an item from the list
l2 = [1,1,2,4.56,True,3.9j,'arshiya']
print(l2)
del(l2[4])
print(l2)
```

```python
l2 = [1,1,2,4.56,True,3.9j,'arshiya']
l3 = [1,1,2,4.56,True,3.9j,'arshiya']
l4 = l2 + l3
print(l4)
```

Tuple

```python
t1 = (3,4,5,6,7)
print(t1)
print(type(t1))
```

```python
#Tuple allows multiple datatypes
t2 = (1,1,2,4.56,True,3.9j,'arshiya')
print(t2)
print(type(t2))
```

```python
#converting a list into tuple
l2 = [1,1,2,4.56,True,3.9j,'arshiya']
t3 = tuple(l2)
print(t3)
print(type(t3))
print(t3[3])
```

```python
#Tuples are immutable, means they are not changeable
#t3.append(9)
print(t3)
print(len(t3))
```

Set

```python
#Set should be represented in curly brackets
s1 = {1,2,3,4,5}
print(s1)
print(type(s1))
#print(s1[0])
#Sets doesnot allow indexing
```

```python
#Convert a list into tuple
l2 = [1,1,2,4.56,True,3.9j,'arshiya']
print(type(l2))
s2 = set(l2)
print(s2)
print(type(s2))
#Sets doesnot allow duplicates, as we can see in the above conversion, the duplicates have been removed
```

Dictionary

Keyvalue Pair

```python
d1 = {'Name': 'Avengers'}
print(d1)
print(type(d1))
```

```python
d2 = {'Name': 'Seetha', 'Age': '34', 'Salary' : '30,000'}
print(d2)
print(type(d2))
print(d2['Name'])
```

```python
#Dictionary doesnot allow duplicate values
d2 = {'Name': 'Seetha', 'Age': '34', 'Salary' : '30,000'}
print(d2)

#d2.append('Job role': 'Associate Engineer')
print(d2)
```
