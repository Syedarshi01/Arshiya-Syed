Simple if else statement

```python
a = 90
b = 89

if (a > b):
  print('a is greater than b')
else:
  print('a is less than b')

```

nested if statement

```python
a = 30
b = 60
c = 200

if (a > b):
  if (a > c):
    print('a is the greatest number')
  else:
    print('c is the greatest number')
else:
  if (b > c):
    print('b is the greatest number')
  else:
    print('c is the greatest number')
```
